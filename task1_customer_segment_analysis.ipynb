```python
customer_metrics = data_full.groupby(['LYLTY_CARD_NBR', 'LIFESTAGE', 'PREMIUM_CUSTOMER']).agg(
    total_spend=('TOT_SALES', 'sum'),
    frequency=('TOT_SALES', 'count'),
    avg_trans_value=('TOT_SALES', 'mean'),
    avg_pack_size=('PACK_SIZE', 'mean')
).reset_index()

customer_metrics.head()
```

```python
segment_summary = customer_metrics.groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER']).agg(
    n_customers=('LYLTY_CARD_NBR', 'count'),
    total_revenue=('total_spend', 'sum'),
    avg_spend=('total_spend', 'mean'),
    avg_frequency=('frequency', 'mean'),
    avg_pack_size=('avg_pack_size', 'mean')
).reset_index().sort_values('total_revenue', ascending=False)

segment_summary
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,6))
sns.barplot(data=segment_summary, x='LIFESTAGE', y='total_revenue', hue='PREMIUM_CUSTOMER')
plt.xticks(rotation=45, ha='right')
plt.title('Total Revenue by Customer Segment')
plt.ylabel('Total Revenue ($)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))
sns.barplot(data=segment_summary, x='LIFESTAGE', y='avg_spend', hue='PREMIUM_CUSTOMER')
plt.xticks(rotation=45, ha='right')
plt.title('Average Spend per Customer by Segment')
plt.xlabel('Lifestage')
plt.ylabel('Average Spend ($)')
plt.show()
```

```python
plt.figure(figsize=(14, 6))
sns.barplot(data=segment_summary, x='LIFESTAGE', y='avg_frequency', hue='PREMIUM_CUSTOMER')
plt.xticks(rotation=45, ha='right')
plt.title('Average Purchase Frequency by Segment')
plt.xlabel('Lifestage')
plt.ylabel('Average Frequency (transactions)')
plt.show()
```

## Summary of Findings

This analysis examined 245,301 chip transactions from 72,637 customers, segmented by
LIFESTAGE (7 categories) and PREMIUM_CUSTOMER tier (Budget/Mainstream/Premium).

### Key Findings

1. **Revenue drivers**: Older Families (Budget) generate the highest total revenue
   ($156,097), driven by high purchase frequency (4.64 transactions/customer). Young
   Singles/Couples (Mainstream) rank 2nd in revenue purely through customer volume
   (7,908 customers), despite lower average spend per customer.

2. **Spend vs. volume**: Average spend per customer is relatively balanced across
   Budget/Mainstream/Premium tiers within each LIFESTAGE, suggesting revenue differences
   are driven more by customer count and frequency than by premium customers spending
   significantly more per visit.

3. **Brand preference**: Kettle is the top-selling brand across every single customer
   segment with no exceptions, making it a universally safe stocking priority. Smiths is
   the dominant second-choice brand in 16 of 21 segments.

4. **Segment outliers**: New Families stands out as a distinct segment — preferring
   Doritos/Pringles over the Smiths/Kettle pattern followed by all other groups. Combined
   with New Families having the lowest revenue and customer count overall, this segment
   may warrant a different marketing approach.

5. **Pack size**: Average pack size is highly consistent across all segments (173g-179g),
   indicating pack size is not a meaningful differentiator of customer purchasing behavior.

### Recommendations to Julia

- **Prioritize Kettle and Smiths** across all store categories and promotional placement,
  as they command consistent demand across every customer segment.
- **Target Older Families and Young Singles/Couples (Mainstream)** as the highest-value
  segments for retention campaigns, given their outsized contribution to total revenue.
- **Consider a tailored strategy for New Families**, given their distinct brand preferences
  (Doritos/Pringles) and currently low engagement — potential for growth if approached
  differently from other segments.
- **Do not prioritize pack size variation by segment** — since preference is consistent
  across all customer types, a standard pack size range (~175g) meets most customer needs.
